# Welcome to **Chroma&Co.py** 🧪
#### *Your Chromatography Assistant*

## Introduction
High Pressure Liquid Chromatography is an essential analytical tool in labs nowadays as it allows to *seperate, identify and quantify* a mixture's components with a high precision. 

Have you ever ran a HPLC to analyse a sample's mixture and wondered if you could skip all the calculations part of your analysis? 

Well you don't have to worry about this with **Chroma&Co.py**, a Python package here to do all those calculations for you !

#### *What are its features ?*

 - Elution order predictions
 - Resolution between two peaks calculations
 - Number of theoretical plates calculations
 - Equivalent height calculations

#### *What do you need to do ?*

**For the elution order** -- Select the type of column used (polar or apolar), provide the number of different molecules you want to elute & input the name of the molecules

**For the Resolution between two peaks** -- Simply provide both the brut retention time and the width of the base of both peaks of interest.

**For the number of theoretical plates** -- Simply provide both the brut retention time and the width of the base of the peak of interest.

**For the equivalent height** -- Simply provide both the brut retention time and the width of the base of the peak of interest and the length of the column.


All those calculations rely on the following formulae :

$$ N(J) = 16 \times \left( \frac{t_R(J)}{w(J)} \right)^2 $$

$$ H(J) = \frac{L}{N(J)} $$

$$ R_S(1,2) = \frac{t_R(2) - t_R(1)}{\frac{1}{2} \times (w(1) + w(2))} $$
